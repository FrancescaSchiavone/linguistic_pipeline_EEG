```python
import numpy as np

def clean_cell_array(x):
    cells = x[0]  # (1,15) → lista di 15 elementi
    cleaned = []
    
    for c in cells:
        # rimuove dimensioni inutili
        c = np.squeeze(c)
        
        # se ancora object → unwrap
        if isinstance(c, np.ndarray) and c.dtype == object:
            c = np.array([np.squeeze(ci) for ci in c])
        
        cleaned.append(c.astype(float))
    
    return cleaned

envelopes = clean_cell_array(x)
for i, e in enumerate(envelopes):
    print(i, e.shape)
```
